# Synthetic ownership data exploration
All records are synthetic. Missing shares and circular structures are data/structural indicators, not criminal labels. Run from the repository root or set ROOT below.

```python
from pathlib import Path
import json, pandas as pd
ROOT = Path.cwd() if (Path.cwd() / 'data').exists() else Path.cwd().parent
dataset = json.loads((ROOT / 'data/sample/demo.json').read_text())
entities = pd.DataFrame(dataset['entities'])
relationships = pd.DataFrame(dataset['relationships'])
display(entities.groupby(['entity_type','jurisdiction']).size().rename('count'))
display(relationships.groupby('control_type').agg(records=('id','count'), missing_percentage=('ownership_percentage', lambda x: x.isna().sum())))
assert entities['external_id'].is_unique
assert entities['synthetic'].all()
```

No supervised metrics can be inferred from these records. Check source IDs, active validity intervals, ownership totals and duplicate stable IDs before feature extraction.
